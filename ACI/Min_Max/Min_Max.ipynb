```python
# code block for CatchGame - Agents Environment
#import itertools
DEBUG_MODE = True  # Set to False to disable print statements
class CatchUpGame:
    def __init__(self, n):
        self.remaininglist = list(range(1, n+1))
        self.game_scores = [0, 0]
        self.current_player = 0
 
    def possible_moves_for_agent(self):
        if self.game_scores[0] == self.game_scores[1]:
            # When the game_scores are equal, all remaining numbers are possible moves
            return [[num] for num in self.remaininglist]
        else:
            return createGame_moves(self.remaininglist, self.game_scores[1-self.current_player]-self.game_scores[self.current_player])
        #return moves
 
    def make_move(self, move):
        for m in move:
            self.remaininglist.remove(m)
            self.game_scores[self.current_player] += m
        self.current_player = 1 - self.current_player
 
    def undo_move(self, move):
        for m in move:
            self.remaininglist.append(m)
            self.game_scores[1-self.current_player] -= m
        self.current_player = 1 - self.current_player
 
    def game_over(self):
        #return len(self.possible_moves_for_agent()) == 0
        return len(self.remaininglist) == 0
```

### Implementation of the Min-Max algorithm and Implementation of the alpha-beta pruning  

```python
def minimax(game, depth, alpha, beta, maximizing_player):
    if depth == 0 or game.game_over():
        if DEBUG_MODE:
            print("******************* Either the game is over or there depth has been reached!!! *******************")
        return evaluateScore(game)
 
    if maximizing_player:
        max_eval = float('-inf')
        for move in game.possible_moves_for_agent():
            game.make_move(move)
            eval = minimax(game, depth-1, alpha, beta, False)
            if DEBUG_MODE:
                print(f"score for a move {move} for maximizing_player is {eval}")
            game.undo_move(move)
            max_eval = max(max_eval, eval)
            alpha = max(alpha, eval)
            if beta <= alpha:
                break
        return max_eval
        
 
    else:
        min_eval = float('inf')
        for move in game.possible_moves_for_agent():
            game.make_move(move)
            eval = minimax(game, depth-1, alpha, beta, True)
            if DEBUG_MODE:
                print(f"score for a move {move} for minimizing_player is {eval}")
            game.undo_move(move)
            min_eval = min(min_eval, eval)
            beta = min(beta, eval)
            if beta <= alpha:
                break
        return min_eval
```

```python
def createGame_moves(available_numbers, target_sum):
    if DEBUG_MODE:
        print(f"Finding moves from remaininglist {available_numbers} for a target sum of {target_sum}")
    # Initialize the list of possible moves
    possible_moves_for_agent = []

    # Sort the available numbers in descending order
    available_numbers.sort(reverse=True)

    # Helper function to find all subsets that sum to the target
    def find_subsets(index, current_sum, current_subset):
        # If the current sum is greater than or equal to the target, add the current subset to the possible moves
        if current_sum >= target_sum:
            possible_moves_for_agent.append(current_subset[:])
            return

        # If we've used all available numbers, return
        if index == len(available_numbers):
            return

        # Include the current number in the subset
        current_subset.append(available_numbers[index])
        find_subsets(index + 1, current_sum + available_numbers[index], current_subset)

        # Exclude the current number from the subset
        current_subset.pop()
        find_subsets(index + 1, current_sum, current_subset)

    # Start the recursive function
    find_subsets(0, 0, [])


    return possible_moves_for_agent
    
```

### Choice and implementation of the Static Evaluation Function.

```python
#Code Block representing static evaluation function.
def evaluateScore(game):
    """
    Static evaluation function that considers potential future moves.
    It computes a potential score by assuming each player makes the best possible move
    from the current state.
    """
    # Calculate the total remaining sum of the numbers
    total_remaining_sum = sum(game.remaininglist)
    
    # Calculate the maximum possible score each player can achieve if they make the best move each time
    max_possible_score_player_1 = game.game_scores[0] + total_remaining_sum
    max_possible_score_player_2 = game.game_scores[1] + total_remaining_sum
    
    # Calculate the difference between the maximum possible game_scores
    score_difference = max_possible_score_player_1 - max_possible_score_player_2

    # Consider potential future moves
    if len(game.remaininglist) > 0:
        # Assume each player makes the best possible move (i.e., chooses the largest available number)
        best_move = max(game.remaininglist)
        if game.current_player == 0:  # If it's player 1's turn
            score_difference += best_move  # Player 1 increases their score
        else:  # If it's player 2's turn
            score_difference -= best_move  # Player 2 decreases the score difference

    return score_difference
    

```

```python
#Code block - Start the game

def play_game(n):
    game = CatchUpGame(n)
    while not game.game_over():
        possible_moves = game.possible_moves_for_agent()
        print(f"Player {game.current_player+1}'s turn. Scores: {game.game_scores}, Remaining: {game.remaininglist}, Possible Moves: {possible_moves}")
        
        if game.current_player == 0:
            move = list(map(int, input("Enter your move (comma-separated numbers): ").split(',')))
            # Check if the entered move is part of the possible moves
            if move not in possible_moves:
                print("Invalid move. Please enter a valid move.")
                continue
        else:
            if DEBUG_MODE:
                print("Previous Player Move ",move)
                print("Possible moves are: ",game.possible_moves_for_agent())
                #print("type(game.possible_moves_for_agent()): ",type(game.possible_moves_for_agent()))
            
            
            moveEval = max(possible_moves,key=lambda x: minimax(game, 3, float('-inf'), float('inf'), True))
            if DEBUG_MODE:
                print (f"move evaluated {moveEval}")
            
            move = moveEval
            
        game.make_move(move)
        print(f"Scores: {game.game_scores}, after a move of: {move} and Remaining: {game.remaininglist}")
 
    
    print(f"Game over! Scores: {game.game_scores}")
    if game.game_scores[0] > game.game_scores[1]:
        print("Player 1 wins!")
    elif game.game_scores[0] < game.game_scores[1]:
        print("Player 2 wins!")
    else:
        print("It's a tie!")


# set the DEBUG_MODE flag to true or false to debug the code.
def main():
    boardSize = input("Enter the initial count of numbers:")
    try:
        size = int(boardSize)
    except ValueError:
        print("The input string is not a valid integer.")
    
    play_game(size)

if __name__ == "__main__":
    main()
```
