### 1.	Define the environment in the following block

List the PEAS decription of the problem here in this markdown block

<b>Performance Measure:</b> The performance of the navigation agent is measured based on two main criteria: maximizing the total points collected and minimizing the number of squares travelled. Each safe cell visited adds 5 points to the agent's score, while passing near water bodies deducts 5 points and passing near flooded areas deducts 3 points.

<b>Environment:</b> The environment is represented as a grid, where each cell can either be a safe area (0), a water body (2000), or a blocked area (1500). The agent can move in four directions (up, down, left, right) but cannot traverse through water bodies or blocked areas. The environment is fully observable as the agent has access to the entire map.

<b>Actuators:</b> The navigation agent can move in four directions: up, down, left, and right. It chooses the next grid cell to move to based on its current position and the surrounding environment.

<b>Sensors:</b> The agent receives sensory input from the environment, allowing it to perceive its current position on the grid and the surrounding cells. It can detect water bodies, blocked areas, and safe cells adjacent to its current position.

<b>Goal:</b> The goal of the navigation agent is to find the safest route to reach the goal state (marked as 1001) while maximizing the total points collected. The agent aims to choose the next grid cell such that it avoids water bodies and blocked areas while passing through as many safe cells as possible. Additionally, it seeks to minimize the number of squares traveled to reach the goal state.

```python
#Code Block : Set Initial State (Must handle dynamic inputs)

import numpy as np

# Define grid dimensions
grid = np.full((8, 8), 0)

# Define function to set initial state
def set_initial_state(grid, initial_state):
    grid[initial_state[0], initial_state[1]] = 1000

# Define function to set goal state
def set_goal_state(grid, goal_state):
    grid[goal_state[0], goal_state[1]] = 1001

# Define function to set water bodies
def set_water_bodies(grid, water_bodies):
    for pos in water_bodies:
        grid[pos[0], pos[1]] = 2000

# Define function to set blockages
def set_blockages(grid, blockages):
    for pos in blockages:
        grid[pos[0], pos[1]] = 1500

# Take input for initial state
initial_state_input = input("Enter initial state coordinates (row column)- separate value by space example- 0 0: ")
initial_state = tuple(map(int, initial_state_input.split()))

# Take input for goal state
goal_state_input = input("Enter goal state coordinates (row column)- separate value by space example- 0 0: ")
goal_state = tuple(map(int, goal_state_input.split()))

# Set initial and goal states
set_initial_state(grid, initial_state)
set_goal_state(grid, goal_state)

# Take input for water bodies
num_water_bodies = int(input("Enter the number of water bodies: "))
water_bodies = []
for _ in range(num_water_bodies):
    water_body_input = input("Enter water body coordinates (row column)- separate value by space example- 0 0: ")
    water_body = tuple(map(int, water_body_input.split()))
    water_bodies.append(water_body)

# Take input for blockages
num_blockages = int(input("Enter the number of blockages: "))
blockages = []
for _ in range(num_blockages):
    blockage_input = input("Enter blockage coordinates (row column)- separate value by space example- 0 0:")
    blockage = tuple(map(int, blockage_input.split()))
    blockages.append(blockage)

# Set water bodies and blockages
set_water_bodies(grid, water_bodies)
set_blockages(grid, blockages)
```

```python
#Code Block : Set the matrix for transition & cost (as relevant for the given problem)

# Define heuristic values for each cell
heuristics = np.zeros((8, 8), dtype=int)
goal_position = goal_state
for i in range(8):
    for j in range(8):
        if grid[i, j] == 0:  # If cell is empty
            heuristics[i, j] = abs(goal_position[0] - i) + abs(goal_position[1] - j)
```

```python
#Code Block : Write function to design the Transition Model/Successor function. Ideally this would be called while search algorithms are implemented

def is_valid_move(i, j):
    return 0 <= i < 8 and 0 <= j < 8 and (i, j) not in water_bodies and grid[i, j] != 1500

def get_neighbors(i, j):
    neighbors = []
    if is_valid_move(i - 1, j):
        neighbors.append((i - 1, j))  # Up
    if is_valid_move(i + 1, j):
        neighbors.append((i + 1, j))  # Down
    if is_valid_move(i, j - 1):
        neighbors.append((i, j - 1))  # Left
    if is_valid_move(i, j + 1):
        neighbors.append((i, j + 1))  # Right
    return neighbors
```

```python
#Code block : Write fucntion to handle goal test (Must handle dynamic inputs). Ideally this would be called while search algorithms are implemented

def is_goal_state(state):
    return state == goal_state
```

### 2.	Definition of Algorithm 1 (Greedy Best First Search)

```python
#Code Block : Function for algorithm 1 implementation - Greedy First Search

def search_algorithm_1():
    start_position = initial_state
    path = [start_position]
    fringe_set = set([start_position])
    closed_set = set()
    expand_count = 0
    total_points = 0

    while not is_goal_state(path[-1]):
        current_position = path[-1]
        neighbors = get_neighbors(*current_position)
        for neighbor in neighbors:
            if neighbor not in closed_set:
                fringe_set.add(neighbor)
        fringe_set.discard(current_position)
        closed_set.add(current_position)
        expand_count += 1

        # Choose the next cell with the lowest heuristic value
        next_position = min(fringe_set, key=lambda x: heuristics[x[0], x[1]])
        path.append(next_position)

        # Compute points for the current move
        i, j = next_position
        if grid[i, j] == 0:  # If the cell is safe
            total_points += 5
        elif grid[i, j] == 2000:  # If the cell is a water body
            total_points -= 5
        elif grid[i, j] == 1500:  # If the cell is blocked
            total_points -= 3

    return path, total_points, expand_count, fringe_set
```

### 3.	Definition of Algorithm 2 (Genetic Algorithm)

```python
#Code Block : Function for algorithm 2 implementation

import random


# Function to initialize a population of paths
def initialize_population(population_size):
    population = []
    for _ in range(population_size):
        path = [initial_state]
        current_position = initial_state
        while not is_goal_state(current_position):
            neighbors = get_neighbors(*current_position)
            valid_neighbors = [neighbor for neighbor in neighbors if neighbor not in path]
            if valid_neighbors:
                next_position = random.choice(valid_neighbors)
                path.append(next_position)
                current_position = next_position
            else:
                break
        population.append(path)
    return population

# Function to evaluate the fitness of a path
def fitness(path):
    total_points = 0
    for position in path:
        i, j = position
        if grid[i, j] == 0:  # If the cell is safe
            total_points += 5
        elif grid[i, j] == 2000:  # If the cell is a water body
            total_points -= 5
        elif grid[i, j] == 1500:  # If the cell is blocked
            total_points -= 3
    return total_points

# Function for crossover (single-point crossover)
def crossover(parent1, parent2):
    crossover_point = random.randint(1, min(len(parent1), len(parent2)) - 1)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2

# Function for mutation (swap mutation)
def mutate(path):
    mutated_path = path.copy()
    if len(mutated_path) > 2:
        mutation_point1, mutation_point2 = random.sample(range(1, len(mutated_path) - 1), 2)
        mutated_path[mutation_point1], mutated_path[mutation_point2] = (
            mutated_path[mutation_point2],
            mutated_path[mutation_point1],
        )
    return mutated_path

# Function for genetic algorithm
def genetic_algorithm(population_size, generations):
    population = initialize_population(population_size)
    unique_cells_expanded = set()
    max_fringe_length = 0

    for generation in range(generations):
        population = sorted(population, key=lambda path: fitness(path), reverse=True)

        # Select the top 50% of paths based on fitness
        selected_paths = population[:population_size // 2]

        # Create offspring through crossover and mutation
        offspring = []
        while len(offspring) < population_size - len(selected_paths):
            parent1, parent2 = random.sample(selected_paths, 2)
            child1, child2 = crossover(parent1, parent2)
            child1 = mutate(child1)
            child2 = mutate(child2)
            offspring.extend([child1, child2])

        # Replace the old population with the new generation
        population = selected_paths + offspring

        # Update metrics
        unique_cells_expanded.update([position for path in selected_paths for position in path])
        max_fringe_length = max(max_fringe_length, len(offspring))

    # Select the best path from the final population
    best_path = max(population, key=lambda path: fitness(path))
    return best_path, fitness(best_path), len(unique_cells_expanded), max_fringe_length
```
